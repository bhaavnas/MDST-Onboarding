Let's compare Michigan to Ohio. Which state has the higher average across all years in the test you chose?

```python
# @@@ 2
mi = states.loc[states["STATE"] == "MICHIGAN", "AVG_MATH_8_SCORE"].mean()
oh = states.loc[states["STATE"] == "OHIO", "AVG_MATH_8_SCORE"].mean()
if (mi > oh):
    print("Michigan has the higher average score.")
else:
    print("Ohio has the higher average score.")
```

Find the average for your chosen test across all states in 2019

```python
# @@@ 3
avg = states.loc[states["YEAR"] == 2019, "AVG_MATH_8_SCORE"].mean()
print(f"The average 8th grade math test score in 2019 was {avg}")
```

```python
# @@@ 4
max_values = states.groupby("STATE")["AVG_MATH_8_SCORE"].max().reset_index()
print(f"The maximum values for 8th grade math scores for each state are: \n{max_values.to_string(index=False)}")
```

<h2> Data Creation </h2>

_Use this space to create train/test data_

```python
from sklearn.model_selection import train_test_split
```

```python
# @@@ 8
states = states.dropna()
test_data = states.sample(n=150, random_state=3)
X = test_data[["STATE_REVENUE"]].iloc[:100]
Y = test_data[["AVG_MATH_8_SCORE"]].iloc[:100]
```

```python
# @@@ 9 

X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=.1, random_state=3)
```

<h2> Prediction </h2>

ML Models [Resource](https://medium.com/@vijaya.beeravalli/comparison-of-machine-learning-classification-models-for-credit-card-default-data-c3cf805c9a5a)

```python
# @@@ 10
# import your sklearn class here
from sklearn import linear_model
```

```python
# @@@ 11
# create your model here
model = linear_model.LinearRegression()
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

## Evaluation

Choose some metrics to evaluate the performance of your model, some of them are mentioned in the tutorial.

```python
# @@@ 12
#mean error
print(f"The mean error is {np.mean(model.predict(X_test)-y_test)}.")

# root mean squared error -- penalizes large errors

print(f"The root mean squared error is {np.mean((model.predict(X_test)-y_test)**2)**0.5}.")
print(states.columns)
```

```python
# @@@ 13

col_name = 'TOTAL_EXPENDITURE'


f = plt.figure(figsize=(12,6))
plt.scatter(X_train[col_name], y_train, color = "red")
plt.scatter(X_train[col_name], model.predict(X_train), color = "green")

plt.legend(['True Training','Predicted Training'])
plt.xlabel(col_name)
plt.ylabel('TOTAL_EXPENDITURE')
plt.title("Model Behavior On Training Set")
```
